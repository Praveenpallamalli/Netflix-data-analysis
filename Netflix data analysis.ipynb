## Data Cleaning
### FIll Missing Director Values

```python
df['director']=df['director'].fillna("unknown")
```

### Observation

- Missing director values were replaced with "Unknown".
- This helps preserve rows without deleting data.

### convert Date Column

```python
df['date_added']=pd.to_datetime(df['date_added'],errors='coerce')
```

### Observation

- date_added column is successfully converted into datetime format.
- Invalid date values are converted into NaT.
- Datetime conversion allows year-wise analysis.

### Remove Duplicate Rows

```python
df=df.drop_duplicates()
```

### Observation

- Duplicate rows were removed from the dataset.
- This improves data quality and prevents repeated records.

## Data Analysis

### Number Of Movies vs Tv Shows

```python
df["type"].value_counts()
```

### Observation

- Netflix contains more movies than TV shows.
- Movies dominate the platform content.

### Top 10 Countries Producing Content

```python
df['country'].value_counts().head(10)
```

### Observation

- The United States contributes the highest amount of Netflix content.
- India and the United Kingdom also contribute significant content.

### Most Common Ratings

```python
df["rating"].value_counts()
```

### Observation

- TV-MA appears to be one of the most common content ratings.
- Netflix contains content for both adults and children.

### Content Added Per Year

```python
df["date_added"].dt.year.value_counts()
```

### Observation

- Netflix content additions increased significantly after 2016.
- The platform experienced rapid growth in later years.

## Visualization

```python
df["type"].value_counts().plot(kind="bar")
plt.title("Movies vs TV Shows")
plt.show()
```

### Observation

- The bar chart clearly shows that movies are more common than TV shows on Netflix.

```python
df['country'].value_counts().head(5).plot(kind='bar')

plt.title("Top 5 Countries Producing Netflix Content")
plt.show()
```

### Observation

- The United States is the leading producer of Netflix content.
- India and other countries also contribute a large number of titles.
- The distribution of content production is uneven across countries.

```python
df["rating"].value_counts().head(10).plot(kind="bar")
plt.title("Top Content Ratings on Netflix")
plt.show()
```

The Netflix dataset contains diverse content from multiple countries and ratings categories. Movies dominate the platform, and content additions increased significantly in recent years.
